2.1 — Load and profile

```python
from pathlib import Path
import sys
import pandas as pd

_src = Path.cwd() / "notebooks" / "src"
if not _src.is_dir():
    _src = Path.cwd().parent / "notebooks" / "src"
sys.path.insert(0, str(_src))
from paths import PROJECT_ROOT
PATH = PROJECT_ROOT / "data/processed/cfpb_v1.csv"

df = pd.read_csv(PATH, low_memory=False)

print("Rows:", f"{len(df):,}")
print("Columns:", list(df.columns))
print(df.dtypes)
df.info(memory_usage="deep")
df.head(3)
```

2.2 — Missing values

```python
missing = df.isna().sum().sort_values(ascending=False)
missing_pct = (missing / len(df) * 100).round(2)
pd.DataFrame({"missing": missing, "pct": missing_pct})
```

2.3 — Date range

```python
df["Date received"] = pd.to_datetime(df["Date received"], errors="coerce")
print("Min:", df["Date received"].min())
print("Max:", df["Date received"].max())
print("Invalid dates:", df["Date received"].isna().sum())
df["Date received"].dt.year.value_counts().sort_index()
```

2.4 — Label column: Issue

```python
issue_counts = df["Issue"].value_counts(dropna=False)
print("Unique issues:", df["Issue"].nunique())
print(issue_counts.head(20))
print("\nTail (rare):")
print(issue_counts.tail(15))
```
